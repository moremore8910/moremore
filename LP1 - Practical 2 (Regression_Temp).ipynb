```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('temperatures.csv')
```

```python
df
```

```python
df.head()
```

```python
# input data
x = df['YEAR']

#output data
y = df['ANNUAL']
```

```python
plt.figure(figsize=(15,5))
plt.title('Temperature Plot of INDIA')
plt.xlabel('Year')
plt.ylabel('Annual Average Temperature')
plt.scatter(x,y)
```

```python
x.shape
```

```python
x = x.values
```

```python
x
```

```python
x = x.reshape(117,1)
```

```python
x
```

```python
x.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regressor = LinearRegression()
```

```python
regressor.fit(x,y)
```

```python
# Value of slope (i.e m) in y= mx + c
regressor.coef_
```

```python
# Value of intercept (i.e c) in y= mx + c
regressor.intercept_
```

```python
regressor.predict([[2094]])
```

```python
predicted = regressor.predict(x)
```

```python
# Values predicted by model which are already known to us in the data
predicted
```

```python
# Actual Values in the data
y
```

```python
import numpy as np
```

```python
# Mean Absolute Error (MAE) method 1
np.mean(abs(y - predicted))
```

```python
# Mean Absolute Error (MAE) method 2
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y,predicted)
```

```python
# Mean Squared Error (MSE) method 1
np.mean(((y - predicted)**2))
```

```python
# Mean Squared Error (MSE) method 2
from sklearn.metrics import mean_squared_error
mean_squared_error(y,predicted)
```

```python
# R-Square or Accuracy method 1
from sklearn.metrics import r2_score
r2_score(y,predicted)
```

```python
# R-Square or Accuracy method 2
regressor.score(x,y)
```

```python
# Linear Regression Model method 1 (Visulaization)
plt.figure(figsize=(5,5))
plt.title('Temperature Plot of INDIA')
plt.xlabel('Year')
plt.ylabel('Annual Average Temperature')
plt.scatter(x, y, label = ' Actual', color = 'red', marker ='.')
plt.plot(x , predicted, label = 'Predicted', color = 'green')
plt.legend()
            
```

```python
# Linear Regression Model method 2
sns.regplot(x='YEAR', y='ANNUAL', data=df)
```
